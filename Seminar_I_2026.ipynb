#### Implementacija AVL stabla

````python
class BinaryNode:

    def __init__ (self, value = None):
        """Create binary node."""
        self.value  = value
        self.left   = None
        self.right  = None
        self.height = 0
        
        
    def computeHeight (self):
        """Compute height of node in BST."""
        height = -1
        if self.left:
            height = max(height, self.left.height)
        if self.right:
            height = max(height, self.right.height)
            
        self.height = height + 1

    def dynamicHeight (self):
        """Compute height of node in BST."""
        height = -1
        if self.left:
            height = max(height, self.left.dynamicHeight())
        if self.right:
            height = max(height, self.right.dynamicHeight())
            
        return height + 1

    def dynamicHeightDifference (self):
        """Compute height difference of node's children in BST."""
        leftTarget = 0
        rightTarget = 0
        if self.left:
            leftTarget = 1 + self.left.dynamicHeight()
        if self.right:
            rightTarget = 1 + self.right.dynamicHeight()
            
        return leftTarget - rightTarget

    def heightDifference (self):
        """Compute height difference of node's children in BST."""
        leftTarget = 0
        rightTarget = 0
        if self.left:
            leftTarget = 1 + self.left.height
        if self.right:
             rightTarget = 1 + self.right.height
                   
        return leftTarget - rightTarget

    def assertAVLProperty (self):
        """Validate AVL property for BST node."""
        if abs(self.dynamicHeightDifference()) > 1:
            return False
        if self.left:
            if not self.left.assertAVLProperty():
                return False
        if self.right:
            if not self.right.assertAVLProperty():
                return False

        return True

    def rotateRight (self):
            """Perform right rotation around given node."""
            newRoot = self.left
            grandson = newRoot.right
            self.left = grandson
            newRoot.right = self
            self.computeHeight()
            return newRoot

    def rotateLeft (self):
        """Perform left rotation around given node."""
        newRoot = self.right
        grandson = newRoot.left
        self.right = grandson
        newRoot.left = self
    
        self.computeHeight()
        return newRoot

    def rotateLeftRight (self):
        """Perform left, then right rotation around given node."""
        child = self.left
        newRoot = child.right
        grand1  = newRoot.left
        grand2  = newRoot.right
        child.right = grand1
        self.left = grand2
    
        newRoot.left = child
        newRoot.right = self
    
        child.computeHeight()
        self.computeHeight()
        
        return newRoot

    def rotateRightLeft (self):
        """Perform right, then left rotation around given node."""
        child = self.right
        newRoot = child.left
        grand1  = newRoot.left
        grand2  = newRoot.right
        child.left = grand2
        self.right = grand1
    
        newRoot.left = self
        newRoot.right = child
    
        child.computeHeight()
        self.computeHeight()
      
        return newRoot

    def compareTo (self, value):
        """
        Returns 0 if equal, negative if smaller and positive if greater.
        Suitable for overriding.
        """
        if self.value == value:
            return 0
        if self.value < value:
            return -1
        return +1
        
    def add (self, val):
        """Adds a new node to BST with value and rebalance as needed."""
        newRoot = self

        # if val <= self.value        
        if self.compareTo(val) >= 0:
            self.left = self.addToSubTree (self.left, val)
            if self.heightDifference() == 2:
                #if val <= self.left.value:
                
                if self.left.compareTo(val) >= 0:
                    newRoot = self.rotateRight()
                else:
                    newRoot = self.rotateLeftRight()
        else:
            self.right = self.addToSubTree (self.right, val)
            if self.heightDifference() == -2:
                #if val > self.right.value:
                if self.right.compareTo(val) < 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()

        newRoot.computeHeight()
        return newRoot

    def newNode (self, val):
        """Return new Binary Node, amenable to subclassing."""
        return BinaryNode(val)

    def addToSubTree (self, parent, val):
        """Add val to parent subtree (if exists) and return root in case it has changed because of rotation."""
        if parent is None:
            return self.newNode(val)

        parent = parent.add(val)
        return parent
           
    def removeFromParent (self, parent, val):
        """Helper method for remove. Ensures proper behavior when removing node that 
        has children."""
        if parent:
            return parent.remove(val)
        return None

    def remove (self, val):
        """
         Remove val from BinaryTree. Works in conjunction with remove
         method in BinaryTree.
        """
        newRoot = self
        rc = self.compareTo(val)
        if rc == 0:
            if self.left is None:
                return self.right

            child = self.left
            while child.right:
                child = child.right
            
            childKey = child.value;
            self.left = self.removeFromParent(self.left, childKey)
            self.value = childKey;

            if self.heightDifference() == -2:
                if self.right.heightDifference() <= 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()
        elif rc > 0:
            self.left = self.removeFromParent(self.left, val)
            if self.heightDifference() == -2:
                if self.right.heightDifference() <= 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()
        else:
            self.right = self.removeFromParent(self.right, val)
            if self.heightDifference() == 2:
                if self.left.heightDifference() >= 0:
                    newRoot = self.rotateRight()
                else:
                    newRoot = self.rotateLeftRight()

        newRoot.computeHeight()
        return newRoot

    def __repr__ (self):
        """Useful debugging function to produce linear tree representation."""
        leftS = ''
        rightS = ''
        if self.left:
            leftS = str(self.left)
        if self.right:
            rightS = str(self.right)
        return "(L:" + leftS + " " + str(self.value) + " R:" + rightS + ")"

    def inorder (self):
        """In order traversal generator of tree rooted at given node."""
        if self.left:
            for n in self.left.inorder():
                yield n

        yield self.value

        if self.right:
            for n in self.right.inorder():
                yield n

class BinaryTree:

    def __init__ (self):
        """Create empty binary tree."""
        self.root = None

    def __repr__ (self):
        if self.root is None:
            return "avl:()"
        return "avl:" + str(self.root)
                
    def newNode (self, value):
        """Return new BinaryNode object. Suitable for overriding."""
        return BinaryNode(value)
    
    def add (self, value):
        """Insert value into proper location in Binary Tree."""
        if self.root is None:
            self.root = self.newNode(value)
        else:
            self.root = self.root.add(value)

    def __contains__ (self, target):
        """Check whether BST contains target value."""
        node = self.root
        while node:
            rc = node.compareTo(target)
            if rc > 0:
                node = node.left
            elif rc < 0:
                node = node.right
            else:
                return True
        return False

           
    def remove (self, val):
        """Remove value from tree."""
        if self.root:
            self.root = self.root.remove(val)

    def __iter__ (self):          # definira iteriranje kroz inorder ispis
        """In order traversal of elements in the tree."""
        if self.root:
            return self.root.inorder()
                        
    def assertAVLProperty (self):
        """Validate AVL property for BST Tree."""
        if self.root:
            return self.root.assertAVLProperty()
        else:
            return True
````

````python
class PlainBSTNode(BinaryNode):
    """Obično BST — bez rotacija."""

    def add(self, val):
        if self.compareTo(val) >= 0:
            self.left = self.addToSubTree(self.left, val)
        else:
            self.right = self.addToSubTree(self.right, val)
        self.computeHeight()
        return self  # nikad ne rotira

    def newNode(self, val):
        return PlainBSTNode(val)


class PlainBST(BinaryTree):
    def newNode(self, value):
        return PlainBSTNode(value)
````

## c) 
Ispisati minimalnu i maksimalnu vrijednost indeksa i pripadnih podataka za te indekse.

````python
def find_min(tree):
    """Najljevlji čvor — minimalna vrijednost indeksa."""
    node = tree.root
    while node.left is not None:
        node = node.left
    return node.value

def find_max(tree):
    """Najdesniji čvor — maksimalna vrijednost indeksa."""
    node = tree.root
    while node.right is not None:
        node = node.right
    return node.value


if __name__ == '__main__':
    for name, tree in [("Obično BST", bst), ("AVL", avl)]:
        mn = find_min(tree)
        mx = find_max(tree)
        print(f"{name}:")
        print(f"  MIN — prezime='{mn[0]}', ime='{mn[1]}'  →  {mn}")
        print(f"  MAX — prezime='{mx[0]}', ime='{mx[1]}'  →  {mx}")
        print()
````

Implementirati metode za ispis broja lijevih rotacija i broja desnih rotacija kod umetanja čvora u AVL stablo. Ispisati broj rotacija pri izgradnji AVL stabla.

````python
_original_rotateLeft = BinaryNode.rotateLeft
_original_rotateRight = BinaryNode.rotateRight
_original_rotateLeftRight = BinaryNode.rotateLeftRight
_original_rotateRightLeft = BinaryNode.rotateRightLeft

class RotationCounter:
    left = 0
    right = 0

    @classmethod
    def reset(cls):
        cls.left = 0
        cls.right = 0

def _counted_rotateLeft(self):
    RotationCounter.left += 1
    return _original_rotateLeft(self)

def _counted_rotateRight(self):
    RotationCounter.right += 1
    return _original_rotateRight(self)

def _counted_rotateLeftRight(self):
    RotationCounter.left += 1
    RotationCounter.right += 1
    return _original_rotateLeftRight(self)

def _counted_rotateRightLeft(self):
    RotationCounter.right += 1
    RotationCounter.left += 1
    return _original_rotateRightLeft(self)

BinaryNode.rotateLeft = _counted_rotateLeft
BinaryNode.rotateRight = _counted_rotateRight
BinaryNode.rotateLeftRight = _counted_rotateLeftRight
BinaryNode.rotateRightLeft = _counted_rotateRightLeft
````

Ispisati stabla po razinama gdje čvorove svake razine treba ispisati u zasebnom retku. S obzirom na veliku količinu podataka dovoljno je ispisati samo prvih 10 razina.

````python
from collections import deque

def print_levels(tree, max_levels=10, max_per_level=20):
    """
    Ispisuje stablo po razinama (BFS). Svaka razina u zasebnom retku.

    max_levels    — koliko razina ispisati (zadatak traži 10)
    max_per_level — limit čvorova po retku da output ne eksplodira
                    (kasnije razine imaju tisuće čvorova)
    """
    if tree.root is None:
        print("Stablo je prazno.")
        return

    queue = deque([tree.root])
    level = 0

    while queue and level < max_levels:
        level_size = len(queue)
        # Skupi prezimena (primarni indeks) na ovoj razini
        keys = []
        for _ in range(level_size):
            node = queue.popleft()
            # node.value je tuple (prezime, ime, spol, datum, mjesto)
            keys.append(node.value[0])
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

        # Skrati ako je predugo
        if len(keys) > max_per_level:
            shown = keys[:max_per_level]
            suffix = f"  ... (+{len(keys) - max_per_level} više)"
        else:
            shown = keys
            suffix = ""

        print(f"Razina {level:2d} ({level_size:6d} čvorova): {', '.join(shown)}{suffix}")
        level += 1


if __name__ == '__main__':
    print("=" * 80)
    print("OBIČNO BST — prvih 10 razina")
    print("=" * 80)
    print_levels(bst, max_levels=10)
    print()
    print("=" * 80)
    print("AVL STABLO — prvih 10 razina")
    print("=" * 80)
    print_levels(avl, max_levels=10)
````

## f)
Implementirati pretraživanje podataka o osobama uporabom implementiranog indeksa i to tako da se za pretraživanje može zadati prezime u obliku prefiksa korištenjem znaka '\*'. Primjerice, može se unijeti cjelovito prezime 'Horvat' ili samo dio odnosno prefiks naziva 'Horv*'. Nakon što je podatak pronađen treba ispisati cijeli zapis o osobi. Usporediti vremena pretrage za obično i balansirano stablo.

````python
def parse_query(query):

    if query.endswith('*'):
        return query[:-1], True
    return query, False


def search_exact(tree, surname):
    results = []

    def _search(node):
        if node is None:
            return
        node_surname = node.value[0]
        if surname < node_surname:
            _search(node.left)
        elif surname > node_surname:
            _search(node.right)
        else:
            _search(node.left)
            results.append(node.value)
            _search(node.right)

    _search(tree.root)
    return results


def search_prefix(tree, prefix):
    results = []

    def _search(node):
        if node is None:
            return
        node_surname = node.value[0]

        if node_surname >= prefix:
            _search(node.left)

        if node_surname.startswith(prefix):
            results.append(node.value)

        if node_surname < prefix or node_surname.startswith(prefix):
            _search(node.right)

    _search(tree.root)
    return results


def search(tree, query):
    key, is_prefix = parse_query(query)
    if is_prefix:
        return search_prefix(tree, key)
    else:
        return search_exact(tree, key)


if __name__ == '__main__':
    test_queries = ['Horvat', 'Horv*', 'Kovač*', 'Žu*', 'Abramović']

    for q in test_queries:
        print(f"Upit: '{q}'")
        print("-" * 60)

        # BST
        t0 = time.perf_counter()
        bst_results = search(bst, q)
        bst_time = time.perf_counter() - t0

        # AVL
        t0 = time.perf_counter()
        avl_results = search(avl, q)
        avl_time = time.perf_counter() - t0

        print(f"  Pronađeno: {len(bst_results):,} zapisa")
        print(f"  BST vrijeme: {bst_time*1000:8.3f} ms")
        print(f"  AVL vrijeme: {avl_time*1000:8.3f} ms")

        # Prikaži prvih 5 rezultata
        if bst_results:
            print(f"  Primjeri:")
            for rec in bst_results[:5]:
                print(f"    {rec}")
            if len(bst_results) > 5:
                print(f"    ... (+{len(bst_results) - 5} više)")
        print()
````

## g)
Implementirati funkciju za traženje osoba u rangu prezimena, primjerice sve osobe od '*Horv*' do '*Ivo*', ili od '*D*' do '*I*'. Rangove treba moći zadati proizvoljno. Usporediti vremena pretrage za obično i balansirano stablo.

````python
def search_range(tree, low, high, inclusive_high_prefix=True):
    results = []

    def in_range(surname):
        if surname < low:
            return False
        if inclusive_high_prefix:
            return surname <= high or surname.startswith(high)
        else:
            return surname <= high

    def _search(node):
        if node is None:
            return
        s = node.value[0]

        if s >= low:
            _search(node.left)

        if in_range(s):
            results.append(node.value)

        if inclusive_high_prefix:
            go_right = s <= high or s.startswith(high)
        else:
            go_right = s <= high
        if go_right:
            _search(node.right)

    _search(tree.root)
    return results


if __name__ == '__main__':
    test_ranges = [
        ('Horv', 'Ivo'),
        ('D', 'I'),
        ('A', 'B'),
        ('Kovač', 'Kovačević'),
        ('Š', 'Ž'),
    ]

    for low, high in test_ranges:
        print(f"Rang: '{low}' .. '{high}'")
        print("-" * 60)

        # BST
        t0 = time.perf_counter()
        bst_results = search_range(bst, low, high)
        bst_time = time.perf_counter() - t0

        # AVL
        t0 = time.perf_counter()
        avl_results = search_range(avl, low, high)
        avl_time = time.perf_counter() - t0

        print(f"  Pronađeno:   {len(bst_results):,} zapisa")
        print(f"  BST vrijeme: {bst_time*1000:8.2f} ms")
        print(f"  AVL vrijeme: {avl_time*1000:8.2f} ms")

        # Prvi i zadnji rezultat (da se vidi raspon)
        if bst_results:
            print(f"  Prvi:        {bst_results[0]}")
            print(f"  Zadnji:      {bst_results[-1]}")
        print()
````

## h)
Sortirati zapise različitim algoritmima sortiranja uzevši kao ključ sortiranja prezime osobe. Ako dvije osobe imaju isto prezime kao drugi ključ sortiranja uzeti ime osobe. Usporediti algoritme po vremenu izvršavanja. Usporedbu vremena sortiranja potrebno je napraviti postepenim povećanjem nasumično izabranog podskupa zapisa koji se sortiraju: veličine 1000, 2000, 5000, 10000, 20000, 50000, 100000. Potrebno je implementirati i usporediti vremena sortiranja za algoritme sortiranja umetanjem, sortiranja spajanjem, brzog sortiranja i sortiranja balansiranim stablom.

````python
import random
import sys

def sort_key(record):
    """Zadana funkcija ključa za zadatak h): (prezime, ime)."""
    return (record[0], record[1])


# --- 1. Insertion sort ---
def insertion_sort(arr, key_fn=sort_key):
    """O(n²). Sporo na velikim podacima."""
    a = list(arr)
    for i in range(1, len(a)):
        current = a[i]
        current_key = key_fn(current)
        j = i - 1
        while j >= 0 and key_fn(a[j]) > current_key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = current
    return a


# --- 2. Merge sort ---
def merge_sort(arr, key_fn=sort_key):
    """O(n log n). Stabilan."""
    a = list(arr)
    _merge_sort_helper(a, 0, len(a) - 1, key_fn)
    return a

def _merge_sort_helper(a, lo, hi, key_fn):
    if lo >= hi:
        return
    mid = (lo + hi) // 2
    _merge_sort_helper(a, lo, mid, key_fn)
    _merge_sort_helper(a, mid + 1, hi, key_fn)
    _merge(a, lo, mid, hi, key_fn)

def _merge(a, lo, mid, hi, key_fn):
    left = a[lo:mid + 1]
    right = a[mid + 1:hi + 1]
    i = j = 0
    k = lo
    while i < len(left) and j < len(right):
        if key_fn(left[i]) <= key_fn(right[j]):
            a[k] = left[i]; i += 1
        else:
            a[k] = right[j]; j += 1
        k += 1
    while i < len(left):
        a[k] = left[i]; i += 1; k += 1
    while j < len(right):
        a[k] = right[j]; j += 1; k += 1


# --- 3. Quick sort ---
def quick_sort(arr, key_fn=sort_key):
    """O(n log n) prosječno. Nasumični pivot da se izbjegne O(n²) na sortiranim podacima.
    Iterativna verzija s eksplicitnim stackom — bez RecursionError-a."""
    a = list(arr)
    stack = [(0, len(a) - 1)]
    while stack:
        lo, hi = stack.pop()
        if lo >= hi:
            continue
        p = _partition(a, lo, hi, key_fn)
        stack.append((lo, p - 1))
        stack.append((p + 1, hi))
    return a

def _partition(a, lo, hi, key_fn):
    pivot_idx = random.randint(lo, hi)
    a[pivot_idx], a[hi] = a[hi], a[pivot_idx]
    pivot_key = key_fn(a[hi])
    i = lo - 1
    for j in range(lo, hi):
        if key_fn(a[j]) <= pivot_key:
            i += 1
            a[i], a[j] = a[j], a[i]
    a[i + 1], a[hi] = a[hi], a[i + 1]
    return i + 1


# --- 4. Tree sort (uporabom AVL-a) ---
def tree_sort(arr, AVLCls=BinaryTree):
    """Umetni sve u AVL stablo `AVLCls` pa in-order obilazak.
    `AVLCls` koristi vlastiti `compareTo` — time se ujedno bira ključ sortiranja."""
    tree = AVLCls()
    for r in arr:
        tree.add(r)
    if tree.root is None:
        return []
    return list(tree.root.inorder())


SIZES = [1000, 2000, 5000, 10000, 20000, 50000, 100000]
INSERTION_MAX_N = 20000

ALGORITHMS = [
    ('Insertion sort', insertion_sort),
    ('Merge sort',     merge_sort),
    ('Quick sort',     quick_sort),
    ('Tree sort',      tree_sort),
]


def measure_sorting(records, sizes, algorithms, key_fn=sort_key, AVLCls=BinaryTree, seed=42):
    """Mjeri vrijeme svakog algoritma na nasumičnim podskupovima zadanih veličina.
    Insertion/merge/quick prima `key_fn`; tree_sort prima `AVLCls`."""
    random.seed(seed)
    results = {name: {} for name, _ in algorithms}

    for n in sizes:
        sample = random.sample(records, n)
        print(f"\n--- n = {n:,} ---")

        for name, fn in algorithms:
            if name == 'Insertion sort' and n > INSERTION_MAX_N:
                print(f"  {name:15s}: preskočeno (n > {INSERTION_MAX_N:,})")
                results[name][n] = None
                continue

            t0 = time.perf_counter()
            if fn is tree_sort:
                fn(sample, AVLCls)
            else:
                fn(sample, key_fn)
            elapsed = time.perf_counter() - t0
            results[name][n] = elapsed
            print(f"  {name:15s}: {elapsed:8.3f} s")

    return results


if __name__ == '__main__':
    timings = measure_sorting(records, SIZES, ALGORITHMS)
````

## i-f) | Pretraga po prefiksu i točnom ključu

Originalne `search_exact` i `search_prefix` iz zadatka f) operiraju nad `node.value[0]`. Definiramo poopćene verzije koje primaju funkciju `key_fn` za izvlačenje ključa iz čvora — semantika algoritma ostaje identična jer su naša stabla i dalje uređena po istom ključu (samo izvlačenje vrijednosti je drugačije).

**Upiti:**
- za datum: točni datumi (`'1965-07-27'`) i prefiksi po godini/mjesecu (`'1995*'`, `'2000-12*'`, `'198*'`)
- za mjesto: točna imena (`'Zagreb'`, `'Osijek'`) i prefiksi (`'Zag*'`, `'Spl*'`, `'A*'`)

Za svaki upit mjerimo i uspoređujemo vremena pretrage na običnom BST-u i AVL stablu.

````python
def search_exact_keyed(tree, query, key_fn):
    """Vraća sve zapise čiji je primarni ključ jednak `query`."""
    results = []
    def _search(node):
        if node is None: return
        node_key = key_fn(node.value)
        if query < node_key:
            _search(node.left)
        elif query > node_key:
            _search(node.right)
        else:
            # Jednak primarni ključ — zapisi s istim ključem mogu biti i lijevo
            # i desno (sortirani po sekundarnom ključu 'ime').
            _search(node.left)
            results.append(node.value)
            _search(node.right)
    _search(tree.root)
    return results


def search_prefix_keyed(tree, prefix, key_fn):
    """Vraća sve zapise čiji primarni ključ počinje znakovima `prefix`."""
    results = []
    def _search(node):
        if node is None: return
        node_key = key_fn(node.value)

        # Idi lijevo ako tamo mogu biti podudaranja (node_key >= prefix)
        if node_key >= prefix:
            _search(node.left)

        if node_key.startswith(prefix):
            results.append(node.value)

        # Idi desno ako tamo mogu biti podudaranja
        if node_key < prefix or node_key.startswith(prefix):
            _search(node.right)
    _search(tree.root)
    return results


def search_keyed(tree, query, key_fn):
    """Dispečer: '*' na kraju upita znači prefiks pretragu."""
    if query.endswith('*'):
        return search_prefix_keyed(tree, query[:-1], key_fn)
    return search_exact_keyed(tree, query, key_fn)


def run_search_benchmark(bst, avl, queries, key_fn, label):
    print("=" * 70)
    print(f"KLJUČ: {label}")
    print("=" * 70)
    for q in queries:
        print(f"\nUpit: '{q}'")
        print("-" * 60)

        t0 = time.perf_counter()
        bst_results = search_keyed(bst, q, key_fn)
        bst_time = time.perf_counter() - t0

        t0 = time.perf_counter()
        avl_results = search_keyed(avl, q, key_fn)
        avl_time = time.perf_counter() - t0

        print(f"  Pronađeno:   {len(bst_results):,} zapisa")
        print(f"  BST vrijeme: {bst_time*1000:8.3f} ms")
        print(f"  AVL vrijeme: {avl_time*1000:8.3f} ms")
        if bst_results:
            print(f"  Primjeri:")
            for rec in bst_results[:3]:
                print(f"    {rec}")
            if len(bst_results) > 3:
                print(f"    ... (+{len(bst_results) - 3} više)")


if __name__ == '__main__':
    datum_queries  = ['1965-07-27', '1995-04-23', '1995*', '2000-12*', '198*']
    mjesto_queries = ['Zagreb', 'Osijek', 'Zag*', 'Spl*', 'A*']

    run_search_benchmark(bst_datum,  avl_datum,  datum_queries,
                         key_fn=lambda v: v[3], label="DATUM ROĐENJA")
    print()
    run_search_benchmark(bst_mjesto, avl_mjesto, mjesto_queries,
                         key_fn=lambda v: v[4], label="MJESTO STANOVANJA")
````

Generaliziramo `search_range` iz zadatka g) da prima `key_fn` za izvlačenje primarnog ključa iz čvora. Algoritam ostaje isti — odsijecanje grana stabla na temelju usporedbe ključa s donjom i gornjom granicom — promijenjeno je samo izvlačenje ključa.

Parametar `inclusive_high_prefix=True` znači da gornju granicu tretiramo i kao prefiks: npr. raspon `'1990-01-01'` do `'1999'` uključuje sve datume iz 1990-ih, a raspon `'D'` do `'I'` sva mjesta čije ime počinje slovima D, E, F, G, H, I.

**Rangovi:**
- za datum: dekadni rasponi godina i konkretni intervali datuma
- za mjesto: rasponi po početnim slovima i konkretni nazivi

````python
def search_range_keyed(tree, low, high, key_fn, inclusive_high_prefix=True):
    """Vraća zapise čiji primarni ključ leži u [low, high]. Po želji,
    high se tretira i kao prefiks (npr. 'D' uključuje sve od 'D...' do 'D~')."""
    results = []

    def in_range(s):
        if s < low:
            return False
        if inclusive_high_prefix:
            return s <= high or s.startswith(high)
        return s <= high

    def _search(node):
        if node is None:
            return
        s = key_fn(node.value)

        if s >= low:
            _search(node.left)

        if in_range(s):
            results.append(node.value)

        if inclusive_high_prefix:
            go_right = s <= high or s.startswith(high)
        else:
            go_right = s <= high
        if go_right:
            _search(node.right)

    _search(tree.root)
    return results


def run_range_benchmark(bst, avl, ranges, key_fn, label):
    print("=" * 70)
    print(f"KLJUČ: {label}")
    print("=" * 70)
    for low, high in ranges:
        print(f"\nRang: '{low}' .. '{high}'")
        print("-" * 60)

        t0 = time.perf_counter()
        bst_results = search_range_keyed(bst, low, high, key_fn)
        bst_time = time.perf_counter() - t0

        t0 = time.perf_counter()
        avl_results = search_range_keyed(avl, low, high, key_fn)
        avl_time = time.perf_counter() - t0

        print(f"  Pronađeno:   {len(bst_results):,} zapisa")
        print(f"  BST vrijeme: {bst_time*1000:8.2f} ms")
        print(f"  AVL vrijeme: {avl_time*1000:8.2f} ms")
        if bst_results:
            print(f"  Prvi:        {bst_results[0]}")
            print(f"  Zadnji:      {bst_results[-1]}")


if __name__ == '__main__':
    datum_ranges = [
        ('1990-01-01', '1999-12-31'),  # cijele 90-te (eksplicitan interval)
        ('2000',       '2009'),        # 2000-te (prefiks)
        ('1961',       '1965'),        # rane 60-te (prefiks)
        ('1985-06-01', '1985-12-31'),  # druga polovica 1985.
    ]
    mjesto_ranges = [
        ('Hor', 'Iva'),
        ('D',   'I'),
        ('A',   'B'),
        ('Š',   'Ž'),
        ('Zag', 'Zag'),  # sva mjesta koja počinju sa 'Zag'
    ]

    run_range_benchmark(bst_datum,  avl_datum,  datum_ranges,
                        key_fn=lambda v: v[3], label="DATUM ROĐENJA")
    print()
    run_range_benchmark(bst_mjesto, avl_mjesto, mjesto_ranges,
                        key_fn=lambda v: v[4], label="MJESTO STANOVANJA")
````

AVL je brži za uže raspone i složenije pretrage s eksplicitnim datumima, dok za velike raspone s prefiksom razlika gotovo nestaje jer dominira `O(k)` skupljanje 100–200k zapisa.

# Analiza algoritama sortiranja

Opažanja iz zadataka **h)** (sortiranje po prezimenu) i **i-h)** (sortiranje po datumu rođenja i mjestu stanovanja). Svi algoritmi sortiranja u našem rješenju primaju opcionalni `key_fn` (odnosno `AVLCls` za tree sort), čime ista implementacija pokriva tri različita indeksna ključa.

## Insertion sort

**Ideja.** Za svaki element s indeksom *i* tražimo njegovo mjesto u već sortiranom prefiksu `arr[0..i-1]`, pomicanjem većih elemenata udesno.

**Složenost.** Najbolji slučaj `O(n)` (već sortirano polje, samo se prolazi jedanput). Prosječni i najgori slučaj `O(n²)`. Stabilan, u mjestu.

```python
def insertion_sort(arr, key_fn=sort_key):
    a = list(arr)
    for i in range(1, len(a)):
        current = a[i]
        current_key = key_fn(current)
        j = i - 1
        while j >= 0 and key_fn(a[j]) > current_key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = current
    return a
```

Zbog kvadratične složenosti algoritam preskačemo za `n > 20.000` — empirijski je vrijeme na `n=20.000` već ~11 s.

## Merge sort

**Ideja.** polje se dijeli na dvije polovice, rekurzivno se sortiraju, pa se spajaju u sortirani niz pomoćnim *merge* korakom.

**Složenost.** `O(n log n)` u svim slučajevima. Prostorna složenost `O(n)` zbog pomoćnih nizova pri spajanju. Stabilan.

Ključna funkcija je *merge*:

```python
def _merge(a, lo, mid, hi, key_fn):
    left = a[lo:mid + 1]
    right = a[mid + 1:hi + 1]
    i = j = 0
    k = lo
    while i < len(left) and j < len(right):
        if key_fn(left[i]) <= key_fn(right[j]):
            a[k] = left[i]; i += 1
        else:
            a[k] = right[j]; j += 1
        k += 1
    while i < len(left):
        a[k] = left[i]; i += 1; k += 1
    while j < len(right):
        a[k] = right[j]; j += 1; k += 1
```

Korištenje `<=` umjesto `<` u usporedbi čuva stabilnost: pri jednakim ključevima element iz lijevog podpolja zadržava prvenstvo.

## Quick sort

**Ideja.** Bira se *pivot* element, polje se particionira tako da su lijevo svi manji ili jednaki, desno svi veći, i postupak se rekurzivno primjenjuje na dijelove.

**Složenost.** Prosječno `O(n log n)`, najgori slučaj `O(n²)` pri lošem odabiru pivota (npr. fiksni desni pivot na već sortiranom polju). Naša implementacija sadrži dvije važne odluke koje to izbjegavaju:

1. **Nasumični pivot.** Pivot biramo `random.randint(lo, hi)`, pa zamijenimo s desnim krajem prije particije. Time se eliminira patološki slučaj na sortiranim ulazima.
2. **Iterativna verzija.** Koristimo eksplicitni `stack` umjesto rekurzije. Time se izbjegava `RecursionError` u Pythonu, koji bi se mogao pojaviti na velikim podacima.

```python
def quick_sort(arr, key_fn=sort_key):
    a = list(arr)
    stack = [(0, len(a) - 1)]
    while stack:
        lo, hi = stack.pop()
        if lo >= hi:
            continue
        p = _partition(a, lo, hi, key_fn)
        stack.append((lo, p - 1))
        stack.append((p + 1, hi))
    return a

def _partition(a, lo, hi, key_fn):
    pivot_idx = random.randint(lo, hi)            # nasumični pivot
    a[pivot_idx], a[hi] = a[hi], a[pivot_idx]
    pivot_key = key_fn(a[hi])
    i = lo - 1
    for j in range(lo, hi):
        if key_fn(a[j]) <= pivot_key:
            i += 1
            a[i], a[j] = a[j], a[i]
    a[i + 1], a[hi] = a[hi], a[i + 1]
    return i + 1
```

## Tree sort

**Ideja.** Sve elemente umetnemo u binarno stablo pretraživanja, a zatim obilaskom dobivamo sortirani niz.

**Složenost.** Sa **AVL** stablom: izgradnja `O(n log n)` (svako od *n* umetanja je `O(log n)`), obilazak `O(n)`. Ukupno `O(n log n)`. S neuravnoteženim BST-om u najgorem slučaju `O(n²)` izgradnja.

```python
def tree_sort(arr, AVLCls=BinaryTree):
    tree = AVLCls()
    for r in arr:
        tree.add(r)
    if tree.root is None:
        return []
    return list(tree.root.inorder())
```

Parametar `AVLCls` omogućava korištenje različitih indeksiranih AVL razreda iz zadatka i) (`DatumAVL`, `MjestoAVL`) — redoslijed sortiranja je tako konzistentan s redoslijedom u indeksu, jer ga oba mehanizma definiraju kroz isti `compareTo`.

U praksi je tree sort ~2× sporiji od merge/quick sorta zbog dodatnog troška AVL rotacija i alokacije čvorova, premda dijeli istu asimptotsku složenost.

````python
import matplotlib.pyplot as plt

all_timings = {
    'Prezime (h)': timings,
    'Datum (i)':   timings_datum,
    'Mjesto (i)':  timings_mjesto,
}

ALGOS_ALL  = ['Insertion sort', 'Merge sort', 'Quick sort', 'Tree sort']
ALGOS_FAST = ['Merge sort', 'Quick sort', 'Tree sort']

COLORS = {
    'Insertion sort': '#d62728',
    'Merge sort':     '#1f77b4',
    'Quick sort':     '#2ca02c',
    'Tree sort':      '#9467bd',
}
MARKERS = {
    'Insertion sort': 'o',
    'Merge sort':     's',
    'Quick sort':     '^',
    'Tree sort':      'D',
}


def plot_grid(scale='log', algos=ALGOS_ALL, title=None):
    """Crta 1x3 grid (po ključu) s odabranim algoritmima i skalom osi.
    scale: 'log' (log–log) ili 'linear'."""
    fig, axes = plt.subplots(1, 3, figsize=(16, 5), sharey=True)
    for ax, (label, t) in zip(axes, all_timings.items()):
        for algo in algos:
            items = sorted((n, v) for n, v in t[algo].items() if v is not None)
            if not items:
                continue
            ns, ts = zip(*items)
            ax.plot(ns, ts, marker=MARKERS[algo], label=algo,
                    color=COLORS[algo], linewidth=2, markersize=7)
        if scale == 'log':
            ax.set_xscale('log')
            ax.set_yscale('log')
        ax.set_xlabel('n  (broj zapisa)')
        ax.set_title(f'Ključ: {label}', fontsize=12)
        ax.grid(True, which='both', alpha=0.3)
        ax.legend(loc='upper left', fontsize=9)
    axes[0].set_ylabel('Vrijeme [s]')
    if title:
        fig.suptitle(title, fontsize=13, y=1.02)
    plt.tight_layout()
    plt.show()


# Tri pogleda na isti dataset
plot_grid(scale='log', algos=ALGOS_ALL,
          title='Vremena sortiranja — log–log skala  (sva 4 algoritma)')

plot_grid(scale='linear', algos=ALGOS_ALL,
          title='Vremena sortiranja — linearna skala  (sva 4 algoritma — vidi dominaciju insertion sorta)')

plot_grid(scale='linear', algos=ALGOS_FAST,
          title='Vremena sortiranja — linearna skala  (samo Merge / Quick / Tree — zoom)')


# --- Tablica faktora rasta ---
print("\nFaktor rasta vremena pri udvostručenju n (n -> 2n):")
print("(vrijednost ~2.0 = O(n log n), ~4.0 = O(n²))\n")
for label, t in all_timings.items():
    print(f"--- {label} ---")
    for algo in ['Merge sort', 'Quick sort', 'Tree sort']:
        sizes = sorted(n for n, v in t[algo].items() if v is not None)
        ratios = []
        for i in range(1, len(sizes)):
            n1, n2 = sizes[i-1], sizes[i]
            if abs(n2 / n1 - 2.0) < 0.6:
                ratios.append(t[algo][n2] / t[algo][n1])
        if ratios:
            avg = sum(ratios) / len(ratios)
            print(f"  {algo:15s}: prosječni faktor rasta = {avg:.2f}")
    print()
````

## Opažanja iz mjerenja

U sva tri grafa (po jedan za svaki indeksni ključ) vidimo dvije jasno odijeljene grupe:

- **Insertion sort** (crveno) — strmiji nagib, vizualno gotovo dvostruki rast vremena po jedinici rasta `n`. To odgovara `O(n²)`.
- **Merge sort, Quick sort, Tree sort** (plavo, zeleno, ljubičasto) — međusobno paralelne linije s blagim nagibom. To odgovara `O(n log n)`.

### Neovisnost o indeksnom ključu

Tri grafa za tri različita ključa (`prezime`, `datum`, `mjesto`) su gotovo identična u oblicima krivulja. To je očekivano: vremenska složenost ovisi o **broju usporedbi i premještanja**, a ne o **tipu vrijednosti** koja se uspoređuje.


AVL stablo je jednako vrijedna struktura za **indeksiranje** (zadaci b)–g)) i za **sortiranje** (zadatak h)) — ali samo to nije dovoljno da bi tree sort bio konkurentan kao opći sortirajući algoritam. U praksi su merge i quick sort značajno brži.

Vrijednost tree sorta otkriva se tek kad **ionako gradimo indeksno stablo** (kao u zadacima i-b)–i-g)): tada sortirani niz dobivamo iz već postojećeg `inorder()` obilaska — bez zasebnog sortirajućeg prolaza.
