#### Implementacija AVL stabla

```python
class BinaryNode:

    def __init__ (self, value = None):
        """Create binary node."""
        self.value  = value
        self.left   = None
        self.right  = None
        self.height = 0
        
        
    def computeHeight (self):
        """Compute height of node in BST."""
        height = -1
        if self.left:
            height = max(height, self.left.height)
        if self.right:
            height = max(height, self.right.height)
            
        self.height = height + 1

    def dynamicHeight (self):
        """Compute height of node in BST."""
        height = -1
        if self.left:
            height = max(height, self.left.dynamicHeight())
        if self.right:
            height = max(height, self.right.dynamicHeight())
            
        return height + 1

    def dynamicHeightDifference (self):
        """Compute height difference of node's children in BST."""
        leftTarget = 0
        rightTarget = 0
        if self.left:
            leftTarget = 1 + self.left.dynamicHeight()
        if self.right:
            rightTarget = 1 + self.right.dynamicHeight()
            
        return leftTarget - rightTarget

    def heightDifference (self):
        """Compute height difference of node's children in BST."""
        leftTarget = 0
        rightTarget = 0
        if self.left:
            leftTarget = 1 + self.left.height
        if self.right:
             rightTarget = 1 + self.right.height
                   
        return leftTarget - rightTarget

    def assertAVLProperty (self):
        """Validate AVL property for BST node."""
        if abs(self.dynamicHeightDifference()) > 1:
            return False
        if self.left:
            if not self.left.assertAVLProperty():
                return False
        if self.right:
            if not self.right.assertAVLProperty():
                return False

        return True

    def rotateRight (self):
            """Perform right rotation around given node."""
            newRoot = self.left
            grandson = newRoot.right
            self.left = grandson
            newRoot.right = self
            self.computeHeight()
            return newRoot

    def rotateLeft (self):
        """Perform left rotation around given node."""
        newRoot = self.right
        grandson = newRoot.left
        self.right = grandson
        newRoot.left = self
    
        self.computeHeight()
        return newRoot

    def rotateLeftRight (self):
        """Perform left, then right rotation around given node."""
        child = self.left
        newRoot = child.right
        grand1  = newRoot.left
        grand2  = newRoot.right
        child.right = grand1
        self.left = grand2
    
        newRoot.left = child
        newRoot.right = self
    
        child.computeHeight()
        self.computeHeight()
        
        return newRoot

    def rotateRightLeft (self):
        """Perform right, then left rotation around given node."""
        child = self.right
        newRoot = child.left
        grand1  = newRoot.left
        grand2  = newRoot.right
        child.left = grand2
        self.right = grand1
    
        newRoot.left = self
        newRoot.right = child
    
        child.computeHeight()
        self.computeHeight()
      
        return newRoot

    def compareTo (self, value):
        """
        Returns 0 if equal, negative if smaller and positive if greater.
        Suitable for overriding.
        """
        if self.value == value:
            return 0
        if self.value < value:
            return -1
        return +1
        
    def add (self, val):
        """Adds a new node to BST with value and rebalance as needed."""
        newRoot = self

        # if val <= self.value        
        if self.compareTo(val) >= 0:
            self.left = self.addToSubTree (self.left, val)
            if self.heightDifference() == 2:
                #if val <= self.left.value:
                
                if self.left.compareTo(val) >= 0:
                    newRoot = self.rotateRight()
                else:
                    newRoot = self.rotateLeftRight()
        else:
            self.right = self.addToSubTree (self.right, val)
            if self.heightDifference() == -2:
                #if val > self.right.value:
                if self.right.compareTo(val) < 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()

        newRoot.computeHeight()
        return newRoot

    def newNode (self, val):
        """Return new Binary Node, amenable to subclassing."""
        return BinaryNode(val)

    def addToSubTree (self, parent, val):
        """Add val to parent subtree (if exists) and return root in case it has changed because of rotation."""
        if parent is None:
            return self.newNode(val)

        parent = parent.add(val)
        return parent
           
    def removeFromParent (self, parent, val):
        """Helper method for remove. Ensures proper behavior when removing node that 
        has children."""
        if parent:
            return parent.remove(val)
        return None

    def remove (self, val):
        """
         Remove val from BinaryTree. Works in conjunction with remove
         method in BinaryTree.
        """
        newRoot = self
        rc = self.compareTo(val)
        if rc == 0:
            if self.left is None:
                return self.right

            child = self.left
            while child.right:
                child = child.right
            
            childKey = child.value;
            self.left = self.removeFromParent(self.left, childKey)
            self.value = childKey;

            if self.heightDifference() == -2:
                if self.right.heightDifference() <= 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()
        elif rc > 0:
            self.left = self.removeFromParent(self.left, val)
            if self.heightDifference() == -2:
                if self.right.heightDifference() <= 0:
                    newRoot = self.rotateLeft()
                else:
                    newRoot = self.rotateRightLeft()
        else:
            self.right = self.removeFromParent(self.right, val)
            if self.heightDifference() == 2:
                if self.left.heightDifference() >= 0:
                    newRoot = self.rotateRight()
                else:
                    newRoot = self.rotateLeftRight()

        newRoot.computeHeight()
        return newRoot

    def __repr__ (self):
        """Useful debugging function to produce linear tree representation."""
        leftS = ''
        rightS = ''
        if self.left:
            leftS = str(self.left)
        if self.right:
            rightS = str(self.right)
        return "(L:" + leftS + " " + str(self.value) + " R:" + rightS + ")"

    def inorder (self):
        """In order traversal generator of tree rooted at given node."""
        if self.left:
            for n in self.left.inorder():
                yield n

        yield self.value

        if self.right:
            for n in self.right.inorder():
                yield n

class BinaryTree:

    def __init__ (self):
        """Create empty binary tree."""
        self.root = None

    def __repr__ (self):
        if self.root is None:
            return "avl:()"
        return "avl:" + str(self.root)
                
    def newNode (self, value):
        """Return new BinaryNode object. Suitable for overriding."""
        return BinaryNode(value)
    
    def add (self, value):
        """Insert value into proper location in Binary Tree."""
        if self.root is None:
            self.root = self.newNode(value)
        else:
            self.root = self.root.add(value)

    def __contains__ (self, target):
        """Check whether BST contains target value."""
        node = self.root
        while node:
            rc = node.compareTo(target)
            if rc > 0:
                node = node.left
            elif rc < 0:
                node = node.right
            else:
                return True
        return False

           
    def remove (self, val):
        """Remove value from tree."""
        if self.root:
            self.root = self.root.remove(val)

    def __iter__ (self):          # definira iteriranje kroz inorder ispis
        """In order traversal of elements in the tree."""
        if self.root:
            return self.root.inorder()
                        
    def assertAVLProperty (self):
        """Validate AVL property for BST Tree."""
        if self.root:
            return self.root.assertAVLProperty()
        else:
            return True
```

```python
class PlainBSTNode(BinaryNode):
    """Obično BST — bez rotacija."""

    def add(self, val):
        if self.compareTo(val) >= 0:
            self.left = self.addToSubTree(self.left, val)
        else:
            self.right = self.addToSubTree(self.right, val)
        self.computeHeight()
        return self  # nikad ne rotira

    def newNode(self, val):
        return PlainBSTNode(val)


class PlainBST(BinaryTree):
    def newNode(self, value):
        return PlainBSTNode(value)
```

```python
def find_min(tree):
    """Najljevlji čvor — minimalna vrijednost indeksa."""
    node = tree.root
    while node.left is not None:
        node = node.left
    return node.value

def find_max(tree):
    """Najdesniji čvor — maksimalna vrijednost indeksa."""
    node = tree.root
    while node.right is not None:
        node = node.right
    return node.value


if __name__ == '__main__':
    for name, tree in [("Obično BST", bst), ("AVL", avl)]:
        mn = find_min(tree)
        mx = find_max(tree)
        print(f"{name}:")
        print(f"  MIN — prezime='{mn[0]}', ime='{mn[1]}'  →  {mn}")
        print(f"  MAX — prezime='{mx[0]}', ime='{mx[1]}'  →  {mx}")
        print()
```

## d)
Implementirati metode za ispis broja lijevih rotacija i broja desnih rotacija kod umetanja čvora u AVL stablo. Ispisati broj rotacija pri izgradnji AVL stabla.

```python
_original_rotateLeft = BinaryNode.rotateLeft
_original_rotateRight = BinaryNode.rotateRight
_original_rotateLeftRight = BinaryNode.rotateLeftRight
_original_rotateRightLeft = BinaryNode.rotateRightLeft

class RotationCounter:
    left = 0
    right = 0

    @classmethod
    def reset(cls):
        cls.left = 0
        cls.right = 0

def _counted_rotateLeft(self):
    RotationCounter.left += 1
    return _original_rotateLeft(self)

def _counted_rotateRight(self):
    RotationCounter.right += 1
    return _original_rotateRight(self)

def _counted_rotateLeftRight(self):
    RotationCounter.left += 1
    RotationCounter.right += 1
    return _original_rotateLeftRight(self)

def _counted_rotateRightLeft(self):
    RotationCounter.right += 1
    RotationCounter.left += 1
    return _original_rotateRightLeft(self)

BinaryNode.rotateLeft = _counted_rotateLeft
BinaryNode.rotateRight = _counted_rotateRight
BinaryNode.rotateLeftRight = _counted_rotateLeftRight
BinaryNode.rotateRightLeft = _counted_rotateRightLeft
```

## e)
Ispisati stabla po razinama gdje čvorove svake razine treba ispisati u zasebnom retku. S obzirom na veliku količinu podataka dovoljno je ispisati samo prvih 10 razina.

```python
from collections import deque

def print_levels(tree, max_levels=10, max_per_level=20):
    """
    Ispisuje stablo po razinama (BFS). Svaka razina u zasebnom retku.

    max_levels    — koliko razina ispisati (zadatak traži 10)
    max_per_level — limit čvorova po retku da output ne eksplodira
                    (kasnije razine imaju tisuće čvorova)
    """
    if tree.root is None:
        print("Stablo je prazno.")
        return

    queue = deque([tree.root])
    level = 0

    while queue and level < max_levels:
        level_size = len(queue)
        # Skupi prezimena (primarni indeks) na ovoj razini
        keys = []
        for _ in range(level_size):
            node = queue.popleft()
            # node.value je tuple (prezime, ime, spol, datum, mjesto)
            keys.append(node.value[0])
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)

        # Skrati ako je predugo
        if len(keys) > max_per_level:
            shown = keys[:max_per_level]
            suffix = f"  ... (+{len(keys) - max_per_level} više)"
        else:
            shown = keys
            suffix = ""

        print(f"Razina {level:2d} ({level_size:6d} čvorova): {', '.join(shown)}{suffix}")
        level += 1


if __name__ == '__main__':
    print("=" * 80)
    print("OBIČNO BST — prvih 10 razina")
    print("=" * 80)
    print_levels(bst, max_levels=10)
    print()
    print("=" * 80)
    print("AVL STABLO — prvih 10 razina")
    print("=" * 80)
    print_levels(avl, max_levels=10)
```

## f)
Implementirati pretraživanje podataka o osobama uporabom implementiranog indeksa i to tako da se za pretraživanje može zadati prezime u obliku prefiksa korištenjem znaka '\*'. Primjerice, može se unijeti cjelovito prezime 'Horvat' ili samo dio odnosno prefiks naziva 'Horv*'. Nakon što je podatak pronađen treba ispisati cijeli zapis o osobi. Usporediti vremena pretrage za obično i balansirano stablo.

```python
def parse_query(query):

    if query.endswith('*'):
        return query[:-1], True
    return query, False


def search_exact(tree, surname):
    results = []

    def _search(node):
        if node is None:
            return
        node_surname = node.value[0]
        if surname < node_surname:
            _search(node.left)
        elif surname > node_surname:
            _search(node.right)
        else:
            _search(node.left)
            results.append(node.value)
            _search(node.right)

    _search(tree.root)
    return results


def search_prefix(tree, prefix):
    results = []

    def _search(node):
        if node is None:
            return
        node_surname = node.value[0]

        if node_surname >= prefix:
            _search(node.left)

        if node_surname.startswith(prefix):
            results.append(node.value)

        if node_surname < prefix or node_surname.startswith(prefix):
            _search(node.right)

    _search(tree.root)
    return results


def search(tree, query):
    key, is_prefix = parse_query(query)
    if is_prefix:
        return search_prefix(tree, key)
    else:
        return search_exact(tree, key)


if __name__ == '__main__':
    test_queries = ['Horvat', 'Horv*', 'Kovač*', 'Žu*', 'Abramović']

    for q in test_queries:
        print(f"Upit: '{q}'")
        print("-" * 60)

        # BST
        t0 = time.perf_counter()
        bst_results = search(bst, q)
        bst_time = time.perf_counter() - t0

        # AVL
        t0 = time.perf_counter()
        avl_results = search(avl, q)
        avl_time = time.perf_counter() - t0

        print(f"  Pronađeno: {len(bst_results):,} zapisa")
        print(f"  BST vrijeme: {bst_time*1000:8.3f} ms")
        print(f"  AVL vrijeme: {avl_time*1000:8.3f} ms")

        # Prikaži prvih 5 rezultata
        if bst_results:
            print(f"  Primjeri:")
            for rec in bst_results[:5]:
                print(f"    {rec}")
            if len(bst_results) > 5:
                print(f"    ... (+{len(bst_results) - 5} više)")
        print()
```

```python
def search_range(tree, low, high, inclusive_high_prefix=True):
    results = []

    def in_range(surname):
        if surname < low:
            return False
        if inclusive_high_prefix:
            return surname <= high or surname.startswith(high)
        else:
            return surname <= high

    def _search(node):
        if node is None:
            return
        s = node.value[0]

        if s >= low:
            _search(node.left)

        if in_range(s):
            results.append(node.value)

        if inclusive_high_prefix:
            go_right = s <= high or s.startswith(high)
        else:
            go_right = s <= high
        if go_right:
            _search(node.right)

    _search(tree.root)
    return results


if __name__ == '__main__':
    test_ranges = [
        ('Horv', 'Ivo'),
        ('D', 'I'),
        ('A', 'B'),
        ('Kovač', 'Kovačević'),
        ('Š', 'Ž'),
    ]

    for low, high in test_ranges:
        print(f"Rang: '{low}' .. '{high}'")
        print("-" * 60)

        # BST
        t0 = time.perf_counter()
        bst_results = search_range(bst, low, high)
        bst_time = time.perf_counter() - t0

        # AVL
        t0 = time.perf_counter()
        avl_results = search_range(avl, low, high)
        avl_time = time.perf_counter() - t0

        print(f"  Pronađeno:   {len(bst_results):,} zapisa")
        print(f"  BST vrijeme: {bst_time*1000:8.2f} ms")
        print(f"  AVL vrijeme: {avl_time*1000:8.2f} ms")

        # Prvi i zadnji rezultat (da se vidi raspon)
        if bst_results:
            print(f"  Prvi:        {bst_results[0]}")
            print(f"  Zadnji:      {bst_results[-1]}")
        print()
```

## h)
Sortirati zapise različitim algoritmima sortiranja uzevši kao ključ sortiranja prezime osobe. Ako dvije osobe imaju isto prezime kao drugi ključ sortiranja uzeti ime osobe. Usporediti algoritme po vremenu izvršavanja. Usporedbu vremena sortiranja potrebno je napraviti postepenim povećanjem nasumično izabranog podskupa zapisa koji se sortiraju: veličine 1000, 2000, 5000, 10000, 20000, 50000, 100000. Potrebno je implementirati i usporediti vremena sortiranja za algoritme sortiranja umetanjem, sortiranja spajanjem, brzog sortiranja i sortiranja balansiranim stablom.

```python
import random
import sys

# Ključ za sortiranje: (prezime, ime). Tuple usporedba u Pythonu radi
# leksikografski pa nam to već daje pravu semantiku.
def sort_key(record):
    return (record[0], record[1])


# --- 1. Insertion sort ---
def insertion_sort(arr):
    """O(n²). Sporo na velikim podacima."""
    a = list(arr)  # kopija da ne mijenjamo original
    for i in range(1, len(a)):
        current = a[i]
        current_key = sort_key(current)
        j = i - 1
        while j >= 0 and sort_key(a[j]) > current_key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = current
    return a


# --- 2. Merge sort ---
def merge_sort(arr):
    """O(n log n). Stabilan."""
    a = list(arr)
    _merge_sort_helper(a, 0, len(a) - 1)
    return a

def _merge_sort_helper(a, lo, hi):
    if lo >= hi:
        return
    mid = (lo + hi) // 2
    _merge_sort_helper(a, lo, mid)
    _merge_sort_helper(a, mid + 1, hi)
    _merge(a, lo, mid, hi)

def _merge(a, lo, mid, hi):
    left = a[lo:mid + 1]
    right = a[mid + 1:hi + 1]
    i = j = 0
    k = lo
    while i < len(left) and j < len(right):
        if sort_key(left[i]) <= sort_key(right[j]):
            a[k] = left[i]; i += 1
        else:
            a[k] = right[j]; j += 1
        k += 1
    while i < len(left):
        a[k] = left[i]; i += 1; k += 1
    while j < len(right):
        a[k] = right[j]; j += 1; k += 1


# --- 3. Quick sort ---
def quick_sort(arr):
    """O(n log n) prosječno. Koristimo nasumični pivot da izbjegnemo
    O(n²) na već sortiranim podacima."""
    a = list(arr)
    # Iterativna verzija s eksplicitnim stackom — izbjegava
    # RecursionError na 100k zapisa
    stack = [(0, len(a) - 1)]
    while stack:
        lo, hi = stack.pop()
        if lo >= hi:
            continue
        p = _partition(a, lo, hi)
        stack.append((lo, p - 1))
        stack.append((p + 1, hi))
    return a

def _partition(a, lo, hi):
    # Nasumični pivot — zamijeni s desnim krajem
    pivot_idx = random.randint(lo, hi)
    a[pivot_idx], a[hi] = a[hi], a[pivot_idx]
    pivot_key = sort_key(a[hi])
    i = lo - 1
    for j in range(lo, hi):
        if sort_key(a[j]) <= pivot_key:
            i += 1
            a[i], a[j] = a[j], a[i]
    a[i + 1], a[hi] = a[hi], a[i + 1]
    return i + 1


# --- 4. Tree sort (uporabom AVL-a) ---
def tree_sort(arr):
    """Umetni sve u AVL pa in-order obilazak.
    O(n log n) izgradnja + O(n) obilazak."""
    tree = BinaryTree()
    for r in arr:
        tree.add(r)

    result = []
    def _inorder(node):
        if node is None:
            return
        _inorder(node.left)
        result.append(node.value)
        _inorder(node.right)
    _inorder(tree.root)
    return result


SIZES = [1000, 2000, 5000, 10000, 20000, 50000, 100000]

ALGORITHMS = [
    ('Insertion sort', insertion_sort),
    ('Merge sort',     merge_sort),
    ('Quick sort',     quick_sort),
    ('Tree sort',      tree_sort),
]

# Inserion sort limitiran na 20.000 jer je na 50.000 ~141s, a na 100.000 ~729s
INSERTION_MAX_N = 20000

def measure_sorting(records, sizes, algorithms, seed=42):
    """
    Vraća dict {algorithm_name: {size: time_in_seconds}}.
    Za svaku veličinu uzima isti nasumični uzorak za sve algoritme.
    """
    random.seed(seed)
    results = {name: {} for name, _ in algorithms}

    for n in sizes:
        # Isti uzorak za sve algoritme — fer usporedba
        sample = random.sample(records, n)
        print(f"\n--- n = {n:,} ---")

        for name, fn in algorithms:
            # Preskoči insertion sort na velikim podacima
            if name == 'Insertion sort' and n > INSERTION_MAX_N:
                print(f"  {name:15s}: preskočeno (n > {INSERTION_MAX_N:,})")
                results[name][n] = None
                continue

            t0 = time.perf_counter()
            sorted_result = fn(sample)
            elapsed = time.perf_counter() - t0
            results[name][n] = elapsed
            print(f"  {name:15s}: {elapsed:8.3f} s")

    return results


if __name__ == '__main__':
    timings = measure_sorting(records, SIZES, ALGORITHMS)
```

```python
import csv
import time
import random
import sys

sys.setrecursionlimit(100_000)

CSV_PATH = "podatkovni_skup_Josipović.csv"

def load_data(path):
    with open(path, encoding="utf-8", newline="") as f:
        return [
            (row[1], row[0], row[2], row[3], row[4])
            for row in csv.reader(f)
        ]


def sort_key_datum(record):
    return (record[3], record[0], record[1])

def sort_key_mjesto(record):
    return (record[4], record[0], record[1])
```

```python
def insertion_sort(arr, key_fn):
    a = list(arr)
    for i in range(1, len(a)):
        current = a[i]
        current_key = key_fn(current)
        j = i - 1
        while j >= 0 and key_fn(a[j]) > current_key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = current
    return a


def merge_sort(arr, key_fn):
    a = list(arr)
    _merge_sort(a, 0, len(a) - 1, key_fn)
    return a

def _merge_sort(a, lo, hi, key_fn):
    if lo >= hi:
        return
    mid = (lo + hi) // 2
    _merge_sort(a, lo, mid, key_fn)
    _merge_sort(a, mid + 1, hi, key_fn)
    _merge(a, lo, mid, hi, key_fn)

def _merge(a, lo, mid, hi, key_fn):
    left = a[lo:mid + 1]
    right = a[mid + 1:hi + 1]

    i = j = 0
    k = lo

    while i < len(left) and j < len(right):
        if key_fn(left[i]) <= key_fn(right[j]):
            a[k] = left[i]; i += 1
        else:
            a[k] = right[j]; j += 1
        k += 1

    while i < len(left):
        a[k] = left[i]; i += 1; k += 1

    while j < len(right):
        a[k] = right[j]; j += 1; k += 1
```

```python
def quick_sort(arr, key_fn):
    a = list(arr)
    stack = [(0, len(a) - 1)]

    while stack:
        lo, hi = stack.pop()
        if lo >= hi:
            continue

        p = partition(a, lo, hi, key_fn)
        stack.append((lo, p - 1))
        stack.append((p + 1, hi))

    return a

def partition(a, lo, hi, key_fn):
    pivot_idx = random.randint(lo, hi)
    a[pivot_idx], a[hi] = a[hi], a[pivot_idx]

    pivot = key_fn(a[hi])
    i = lo - 1

    for j in range(lo, hi):
        if key_fn(a[j]) <= pivot:
            i += 1
            a[i], a[j] = a[j], a[i]

    a[i + 1], a[hi] = a[hi], a[i + 1]
    return i + 1


class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class SimpleBST:
    def __init__(self):
        self.root = None

    def add(self, value):
        self.root = self._add(self.root, value)

    def _add(self, node, value):
        if node is None:
            return Node(value)

        if value <= node.value:
            node.left = self._add(node.left, value)
        else:
            node.right = self._add(node.right, value)

        return node


def tree_sort(arr, key_fn):
    tree = SimpleBST()

    for r in arr:
        tree.add((key_fn(r), r))

    result = []

    def inorder(node):
        if node is None:
            return
        inorder(node.left)
        result.append(node.value[1])
        inorder(node.right)

    inorder(tree.root)
    return result
```

```python
SIZES = [1000, 2000, 5000, 10000, 20000, 50000]
INSERTION_MAX = 20000

ALGORITHMS = [
    ("Insertion", insertion_sort),
    ("Merge", merge_sort),
    ("Quick", quick_sort),
    ("Tree", tree_sort),
]


def measure(records, key_fn, label):
    print(f"\n===== KLJUČ: {label} =====")

    for n in SIZES:
        sample = random.sample(records, n)
        print(f"\n--- n = {n} ---")

        for name, fn in ALGORITHMS:
            if name == "Insertion" and n > INSERTION_MAX:
                print(f"{name:10s}: preskočeno")
                continue

            t0 = time.perf_counter()
            fn(sample, key_fn)
            t = time.perf_counter() - t0

            print(f"{name:10s}: {t:.4f} s")
```

## Dodatna zadaća 2: Red-Black Stablo (+3 boda)
Implementacija Red-Black stabla s rotacijama i rebalansiranjem za sve operacije.

```python
class Color:
    RED = 0
    BLACK = 1

class RBNode:
    def __init__(self, value, color=Color.RED):
        self.value = value
        self.color = color
        self.left = None
        self.right = None
        self.parent = None

    def uncle(self):
        if self.parent is None or self.parent.parent is None:
            return None
        if self.parent == self.parent.parent.left:
            return self.parent.parent.right
        else:
            return self.parent.parent.left

    def sibling(self):
        if self.parent is None:
            return None
        if self == self.parent.left:
            return self.parent.right
        else:
            return self.parent.left


class RedBlackTree:
    def __init__(self):
        self.root = None
        self.rotation_count = 0

    def add(self, value):
        if self.root is None:
            self.root = RBNode(value, Color.BLACK)
            return

        self._insert(value)

    def _insert(self, value):
        node = RBNode(value)
        current = self.root
        parent = None

        while current:
            parent = current
            if value <= current.value:
                current = current.left
            else:
                current = current.right

        node.parent = parent
        if value <= parent.value:
            parent.left = node
        else:
            parent.right = node

        self._fix_insert(node)

    def _fix_insert(self, node):
        while node.parent and node.parent.color == Color.RED:
            if node.parent == node.parent.parent.left:
                uncle = node.uncle()
                if uncle and uncle.color == Color.RED:
                    node.parent.color = Color.BLACK
                    uncle.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    node = node.parent.parent
                else:
                    if node == node.parent.right:
                        node = node.parent
                        self._rotate_left(node)
                    node.parent.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    self._rotate_right(node.parent.parent)
            else:
                uncle = node.uncle()
                if uncle and uncle.color == Color.RED:
                    node.parent.color = Color.BLACK
                    uncle.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    node = node.parent.parent
                else:
                    if node == node.parent.left:
                        node = node.parent
                        self._rotate_right(node)
                    node.parent.color = Color.BLACK
                    node.parent.parent.color = Color.RED
                    self._rotate_left(node.parent.parent)

        self.root.color = Color.BLACK

    def _rotate_left(self, node):
        self.rotation_count += 1
        right_child = node.right
        node.right = right_child.left
        if right_child.left:
            right_child.left.parent = node

        right_child.parent = node.parent
        if node.parent is None:
            self.root = right_child
        elif node == node.parent.left:
            node.parent.left = right_child
        else:
            node.parent.right = right_child

        right_child.left = node
        node.parent = right_child

    def _rotate_right(self, node):
        self.rotation_count += 1
        left_child = node.left
        node.left = left_child.right
        if left_child.right:
            left_child.right.parent = node

        left_child.parent = node.parent
        if node.parent is None:
            self.root = left_child
        elif node == node.parent.right:
            node.parent.right = left_child
        else:
            node.parent.left = left_child

        left_child.right = node
        node.parent = left_child

    def inorder(self):
        result = []
        def _traverse(node):
            if node is None:
                return
            _traverse(node.left)
            result.append(node.value)
            _traverse(node.right)
        _traverse(self.root)
        return result
```
